```python
on_time_or_early = orders[orders['delay_days'] <= 0].shape[0]

delayed = orders[orders['delay_days'] > 0].shape[0]

print("On-time or early:", on_time_or_early)
print("Delayed:", delayed)
```

```python
total_delivery_orders = on_time_or_early + delayed

on_time_pct = (
    on_time_or_early / total_delivery_orders
) * 100

delayed_pct = (
    delayed / total_delivery_orders
) * 100

print("On-time or early %:", round(on_time_pct, 2))
print("Delayed %:", round(delayed_pct, 2))
```

```python
average_delay = orders.loc[
    orders['delay_days'] > 0,
    'delay_days'
].mean()

print("Average delay:", round(average_delay, 2), "days")
```

```python
max_delay = orders['delay_days'].max()

print("Maximum delay:", max_delay, "days")
```

```python
most_delayed_order = orders.loc[
    orders['delay_days'].idxmax(),
    ['order_id', 'delay_days',
     'order_delivered_customer_date',
     'order_estimated_delivery_date']
]

print(most_delayed_order)
```
